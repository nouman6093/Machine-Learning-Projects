```python
#dataset = https://www.kaggle.com/datasets/avinashskumar/kyphosisa-fata-disease-for-children

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
df = pd.read_csv('Datasets/kyphosis.csv')
```

```python
x = df.drop(columns = ['Kyphosis'])
y = df['Kyphosis']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3)  #i just found out that sometimes not writing random state increases performance
```

# **Decision Tree**

```python
dtree = DecisionTreeClassifier()
```

```python
dtree.fit(x_train, y_train)
```

```python
predictions = dtree.predict(x_test)
```

```python
print(classification_report(y_test, predictions))
```

```python
sns.heatmap(confusion_matrix(y_test, predictions), annot = True)
```

# **Random Forest**

```python
rfc = RandomForestClassifier(n_estimators = 200)
```

```python
rfc.fit(x_train, y_train)
```

```python
predictions2 = rfc.predict(x_test)
```

```python
print(classification_report(y_test, predictions2))
```

```python
sns.heatmap(confusion_matrix(y_test, predictions2), annot = True)
```
