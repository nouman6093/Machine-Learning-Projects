```python
#dataset = https://www.kaggle.com/datasets/nareshbhat/health-care-data-set-on-heart-attack-possibility

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
import kagglehub
from kagglehub import KaggleDatasetAdapter
```

```python
kagglehub.dataset_download('nareshbhat/health-care-data-set-on-heart-attack-possibility')
```

```python
df = pd.read_csv(r'C:\\Users\\Lenovo\\.cache\\kagglehub\\datasets\\nareshbhat\\health-care-data-set-on-heart-attack-possibility\\versions\\1\heart.csv')
```

```python
df.head()
```

```python
x = df.drop(columns = ['target'])
y = df['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state = 42)
```

```python
model = LogisticRegression(max_iter = 1000, class_weight = 'balanced')
```

```python
model.fit(x_train, y_train)
```

```python
predictions = model.predict(x_test)
```

```python
print(classification_report(y_test, predictions))
```

```python
sns.heatmap(confusion_matrix(y_test, predictions), annot = True, fmt = 'd', cmap = 'Blues')
```
